```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")

print("Path to dataset files:", path)
```

```python
import cudf
df = cudf.read_csv("/home/cjoshi1412/.cache/kagglehub/datasets/mlg-ulb/creditcardfraud/versions/3/creditcard.csv")
```

```python
df
```

```python
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
df.shape
```

```python
y = df['Class']
```

```python
X = df.drop(columns=['Class'])
```

```python
from cuml.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42, stratify=y)
```

```python
print("Train X shape:",X_train.shape)
print("Train y shape:",y_train.shape)
```

```python
X_train, X_val, y_train, y_val = train_test_split(X_train,y_train, test_size=0.2,random_state=42,stratify=y_train)
```

```python
from cuml.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
#SVC
```

```python
X_t = scaler.fit_transform(X_train)
```

```python
from cuml.svm import SVC
clf = SVC()
clf.fit(X_t, y_train)
```

```python
print(clf.n_iter_)
```

```python
X_v = scaler.transform(X_val)
```

```python
y_val_pred_scores = clf.decision_function(X_v)
```

```python
type(y_val_pred_scores)
```

```python
type(y_val)
```

```python
from sklearn.metrics import average_precision_score
```

```python
auprc = average_precision_score(y_val.astype(int).to_numpy(), y_val_pred_scores.to_numpy())
baseline = y_val.astype(int).mean()  # or sum(y_val==1)/len(y_val)
print(f"Baseline (random) AUPRC: {baseline:.4f}")
print(f"Your model's AUPRC:      0.8165")
print(f"Lift over baseline:      {0.8165/baseline:.1f}x")
```

```python
from sklearn.metrics import precision_recall_curve
import matplotlib.pyplot as plt
precisions, recalls, thresholds = precision_recall_curve(y_val.astype(int).to_numpy(), y_val_pred_scores.to_numpy())
```

```python
def plot_precision_recall_vs_threshold(precisions, recalls, thresholds):
    plt.plot(thresholds, precisions[:-1], "b--", label="Precision")
    plt.plot(thresholds, recalls[:-1], "g-", label="Recall")
    plt.xlabel("Decision Threshold")
    plt.ylabel("Score")
    plt.title("Precision & Recall vs. Decision Threshold")
    plt.legend()
    plt.grid(alpha=0.3)
```

```python
plot_precision_recall_vs_threshold(precisions, recalls, thresholds)
plt.show()
```

```python
#class weight balanced
```

```python
clf_2 = SVC(class_weight="balanced")
clf_2.fit(X_t, y_train)
```

```python
y_val_pred_scores_2 = clf_2.decision_function(X_v)
auprc = average_precision_score(y_val.astype(int).to_numpy(), y_val_pred_scores_2.to_numpy())
baseline = y_val.astype(int).mean()  # or sum(y_val==1)/len(y_val)
print(f"Baseline (random) AUPRC: {baseline:.4f}")
print(f"Your model's AUPRC:      {auprc}")
print(f"Lift over baseline:      {auprc/baseline:.1f}x")
```

```python
precisions, recalls, thresholds = precision_recall_curve(y_val.astype(int).to_numpy(), y_val_pred_scores_2.to_numpy())
plot_precision_recall_vs_threshold(precisions, recalls, thresholds)
plt.show()
```

```python
import numpy as np
import cupy as cp

# Convert to numpy, handling either cudf Series or cupy array
def to_numpy(x):
    if hasattr(x, "to_numpy"):       # cudf Series
        return x.to_numpy()
    elif isinstance(x, cp.ndarray):  # cupy array
        return cp.asnumpy(x)
    return np.asarray(x)

y_val_np = to_numpy(y_val.astype(int))
scores_np = to_numpy(y_val_pred_scores_2)

print("Fraud scores:  ", scores_np[y_val_np == 1].min(), scores_np[y_val_np == 1].max())
print("Legit scores:  ", scores_np[y_val_np == 0].min(), scores_np[y_val_np == 0].max())
# Check if solver even converged
print(clf_2.n_iter_)
```

```python
import xgboost as xgb
clf_xgb = xgb.XGBClassifier(
    n_estimators=200,
    eval_metric='aucpr',  # XGBoost can optimize/monitor AUPRC directly
    device='cuda'
)
clf_xgb.fit(X_t, y_train.astype(int))
scores_xgb = clf_xgb.predict_proba(X_v)[:, 1]
```

```python
#with class weight balanced
```

```python
clf_xgb_2 = xgb.XGBClassifier(
    n_estimators=200,
    scale_pos_weight=(y_train.astype(int)==0).sum() / (y_train.astype(int)==1).sum(),  # built-in imbalance handling
    eval_metric='aucpr',  # XGBoost can optimize/monitor AUPRC directly
    device='cuda'
)
clf_xgb_2.fit(X_t, y_train.astype(int))
scores_xgb_2 = clf_xgb_2.predict_proba(X_v)[:, 1]
```

```python
auprc = average_precision_score(y_val.astype(int).to_numpy(), scores_xgb_2)
baseline = y_val.astype(int).mean()  # or sum(y_val==1)/len(y_val)
print(f"Baseline (random) AUPRC: {baseline:.4f}")
print(f"Your model's AUPRC:      {auprc}")
print(f"Lift over baseline:      {auprc/baseline:.1f}x")
```

```python
precisions, recalls, thresholds = precision_recall_curve(y_val.astype(int).to_numpy(), scores_xgb_2)
plot_precision_recall_vs_threshold(precisions, recalls, thresholds)
plt.show()
```

```python
#Final comparison
```

```python
import pandas as pd
auprc_results = {
    "SVC (unweighted)": 0.8165,
    "SVC (balanced)": 0.4795,
    "Logistic Regression (unweighted)": 0.8154,
    "Logistic Regression (balanced)": 0.8485,
    "Random Forest (unweighted)": 0.8590,
    "Random Forest (balanced)": 0.8818,
    "XGBoost (unweighted)": 0.8336,
    "XGBoost (balanced)": 0.8844
}

baseline_auprc = y_val.astype(int).to_numpy().mean()  # random baseline (fraud rate)

# --- Build table ---
df_results = pd.DataFrame([
    {"Model": name, "AUPRC": score, "Lift over baseline": (score / baseline_auprc) if score is not None else None}
    for name, score in auprc_results.items()
])
df_results = df_results.dropna(subset=["AUPRC"]).sort_values("AUPRC", ascending=False).reset_index(drop=True)

print(df_results.to_string(index=False))

# --- Visualize ---
fig, ax = plt.subplots(figsize=(8, 5))

colors = ["#4C72B0" if "Isolation" not in m else "#C44E52" for m in df_results["Model"]]
bars = ax.barh(df_results["Model"], df_results["AUPRC"], color=colors)

ax.axvline(baseline_auprc, color="gray", linestyle="--", linewidth=1, label=f"Random baseline ({baseline_auprc:.4f})")

# Add value labels on bars
for bar, score in zip(bars, df_results["AUPRC"]):
    ax.text(bar.get_width() + 0.01, bar.get_y() + bar.get_height()/2,
             f"{score:.4f}", va="center", fontsize=9)

ax.set_xlabel("AUPRC")
ax.set_title("Model Comparison: AUPRC on Credit Card Fraud Detection")
ax.set_xlim(0, max(df_results["AUPRC"]) * 1.15)
ax.invert_yaxis()  # highest score at top
ax.legend(loc="lower right")
ax.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.savefig("auprc_model_comparison.png", dpi=150)
plt.show()
```

```python
df_results.to_csv("auprc_model_comparison.csv")
```
